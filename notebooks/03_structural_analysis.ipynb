```python
from pymatgen.core import Structure
from pymatgen.symmetry.analyzer import SpacegroupAnalyzer

import pandas as pd
import numpy as np
```

```python
## loading the data

hbn = Structure.from_file("..\\data\\h-BN.cif")

si = Structure.from_file("..\\data\\sillicon.cif")
```

```python
## basic structural properties

print("h-BN: Structure")
print(hbn)

print("\nSi: Structure")
print(si)
```

```python
## Composition analysis

print("h-BN: Composition")
print(hbn.composition)

print("\nSi: Composition")
print(si.composition)
```

```python
## number of atoms

print("h-BN: Number of atoms", len(hbn))
print("Si: Number of atoms", len(si))   
```

```python
## lattice parameters

print("h-BN: Lattice parameters")
print(hbn.lattice.abc)

print("\nSi: Lattice parameters")
print(si.lattice.abc)
```

```python
## lattice angles

print("h-BN: Lattice angles")
print(hbn.lattice.angles)

print("\nSi: Lattice angles")
print(si.lattice.angles)
```

```python
## Unit cell volume 

print("h-BN: Unit cell volume", hbn.lattice.volume)
print("Si: Unit cell volume", si.lattice.volume)
```

```python
## density analysis

print("h-BN: Density")
print(hbn.density)

print("\nSi: Density")
print(si.density)
```

```python
## Space group analysis

hbn_sga = SpacegroupAnalyzer(hbn)
si_sga = SpacegroupAnalyzer(si)


print("h-BN: Space group symbol")
print(hbn_sga.get_space_group_symbol())

print("\nSi: Space group symbol")
print(si_sga.get_space_group_symbol())
```

```python
##  Crystal system analysis

print("h-BN: Crystal system")
print(hbn_sga.get_crystal_system())

print("\nSi: Crystal system")
print(si_sga.get_crystal_system())
```

```python
## Site information

print("h-BN: Sites")

for i, site in enumerate(hbn):
    print(i,site)
```

```python
print("\nSi sites")
for i, site in enumerate(si):
    print(i,site)
```

```python
## Bond length analysis

print("B-N bond lengths in h-BN")
print(hbn.get_distance(0, 2))

print("\nSi-Si bond lengths in Si")
print(si.get_distance(0, 1))
```

```python
## nearest neighbor analysis of h-BN

neighbors_hbn = hbn.get_neighbors(hbn[0], 2.0)

print("h-BN neighbors:\n")


for n in neighbors_hbn:
    print(n)
```

```python
## nearest neighbor analysis of Si


neighbors_si = si.get_neighbors(si[0], 3.0)

print("Si neighbors:\n")
for n in neighbors_si:
    print(n)
```

```python
## Coordination number

print("h-BN: Coordination number")
print(len(neighbors_hbn))

print("\nSi: Coordination number")
print(len(neighbors_si))
```

```python
## Fractional coordinates of h-BN 

print("h-BN: Fractional coordinates")

for site in hbn:
    print(site.frac_coords)
```

```python
## Fractional coordinates of Si

print("\nSi: Fractional coordinates")

for site in si:
    print(site.frac_coords)
```

```python
## cartesian coordinates of h-BN

print("\nh-BN: Cartesian coordinates")

for site in hbn:
    print(site.coords)
```

```python
## Cartesian coordinates of Si

print("\nSi: Cartesian coordinates")

for site in si:
    print(site.coords)
```

```python
## Structural comparision table 

comparison_data = {
    "Property": ["Composition",
                 "Number of atoms",
                "Lattice parameters (a,b,c)", 
                "Lattice angles (alpha, beta, gamma)",
                "Unit cell volume", 
                "Density",
                "Space group symbol", 
                "Crystal system",
                "Coordination number"],


    "h-BN": [hbn.composition.formula,
            len(hbn),
            hbn.lattice.abc,
            hbn.lattice.angles,
            hbn.lattice.volume,
            hbn.density,
            hbn_sga.get_space_group_symbol(),
            hbn_sga.get_crystal_system(),
            len(neighbors_hbn)],

    "Si": [si.composition.formula,
            len(si),
            si.lattice.abc,
            si.lattice.angles, 
            si.lattice.volume, si.density, 
            si_sga.get_space_group_symbol(), 
            si_sga.get_crystal_system(),
            len(neighbors_si)]
}

df = pd.DataFrame(comparison_data)

df
```
